```python

# ============================================================
# SRM INSTITUTE OF SCIENCE AND TECHNOLOGY
# COURSE: 21CSC301T - FORMAL LANGUAGE AND AUTOMATA
# CLA-2 ASSIGNMENT 2
# FINITE AUTOMATA FOR REAL-TIME INPUT VALIDATION
#
# STUDENT NAME: PARIMI GOWTHAM
# REGISTRATION NO: RA2411026050033
#
# APPLICATIONS:
# 1. Indian Mobile Number Validator
# 2. University Course Code Validator
# 3. Hexadecimal Colour Code Validator
# ============================================================

# =========================
# 1. IMPORT LIBRARIES
# =========================

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from dataclasses import dataclass
from typing import Dict, Tuple, List, Callable

print("=" * 65)
print("FORMAL LANGUAGE AND AUTOMATA - ASSIGNMENT 2")
print("STUDENT: PARIMI GOWTHAM")
print("REGISTRATION NO: RA2411026050033")
print("=" * 65)


# =========================
# 2. DFA ENGINE
# =========================

@dataclass
class DFA:

    name: str
    states: List[str]
    alphabet: List[str]
    start_state: str
    final_states: set
    trap_state: str
    transitions: Dict[Tuple[str, str], str]
    classifier: Callable

    def classify(self, char):
        return self.classifier(char)

    def transition(self, state, char):

        symbol = self.classify(char)

        return self.transitions.get(
            (state, symbol),
            self.trap_state
        )

    def run(self, input_string):

        current_state = self.start_state

        state_trace = [current_state]

        for char in input_string:

            current_state = self.transition(
                current_state,
                char
            )

            state_trace.append(current_state)

        accepted = current_state in self.final_states

        return accepted, state_trace

    def display_transition_table(self):

        rows = []

        for state in self.states:

            row = {"State": state}

            for symbol in self.alphabet:

                row[symbol] = self.transitions.get(
                    (state, symbol),
                    self.trap_state
                )

            row["Final State"] = (
                "YES" if state in self.final_states else "NO"
            )

            rows.append(row)

        return pd.DataFrame(rows)

    def trace_input(self, input_string):

        current_state = self.start_state

        rows = []

        for position, char in enumerate(input_string, 1):

            symbol = self.classify(char)

            next_state = self.transition(
                current_state,
                char
            )

            rows.append({
                "Position": position,
                "Character": char,
                "Symbol Class": symbol,
                "Current State": current_state,
                "Next State": next_state
            })

            current_state = next_state

        accepted = current_state in self.final_states

        return pd.DataFrame(rows), accepted, current_state


def create_dfa(
    name,
    states,
    alphabet,
    start,
    finals,
    trap,
    transitions,
    classifier
):

    states = list(states)

    if trap not in states:
        states.append(trap)

    transitions = dict(transitions)

    # Trap state loops for every input class
    for symbol in alphabet:

        transitions[(trap, symbol)] = trap

    return DFA(
        name,
        states,
        alphabet,
        start,
        set(finals),
        trap,
        transitions,
        classifier
    )


# =========================
# 3. DFA DIAGRAM GENERATOR
# =========================

def draw_dfa(dfa, title):

    normal_states = [
        state for state in dfa.states
        if state != dfa.trap_state
    ]

    count = len(normal_states)

    fig, ax = plt.subplots(
        figsize=(max(12, count * 1.2), 4)
    )

    positions = {
        state: (i * 1.5, 1.2)
        for i, state in enumerate(normal_states)
    }

    trap_position = (
        max(0, (count - 1) * 1.5 / 2),
        -0.8
    )

    positions[dfa.trap_state] = trap_position

    # Draw normal transitions
    grouped = {}

    for (source, symbol), destination in dfa.transitions.items():

        if source == dfa.trap_state:
            continue

        if destination == dfa.trap_state:
            continue

        grouped.setdefault(
            (source, destination), []
        ).append(symbol)

    for (source, destination), labels in grouped.items():

        x1, y1 = positions[source]
        x2, y2 = positions[destination]

        ax.annotate(
            "",
            xy=(x2 - 0.25, y2),
            xytext=(x1 + 0.25, y1),
            arrowprops=dict(
                arrowstyle="->",
                linewidth=1.5
            )
        )

        ax.text(
            (x1 + x2) / 2,
            y1 + 0.2,
            ",".join(labels),
            ha="center",
            fontsize=9
        )

    # Draw transitions to trap state
    for (source, symbol), destination in dfa.transitions.items():

        if source != dfa.trap_state and destination == dfa.trap_state:

            x, y = positions[source]
            tx, ty = trap_position

            ax.annotate(
                "",
                xy=(tx, ty + 0.2),
                xytext=(x, y - 0.2),
                arrowprops=dict(
                    arrowstyle="->",
                    linewidth=0.6,
                    alpha=0.25,
                    color="red",
                    connectionstyle="arc3,rad=0.12"
                )
            )

    # Draw states
    for state, (x, y) in positions.items():

        circle = Circle(
            (x, y),
            0.25,
            facecolor="white",
            edgecolor="navy",
            linewidth=2
        )

        ax.add_patch(circle)

        if state in dfa.final_states:

            ax.add_patch(
                Circle(
                    (x, y),
                    0.20,
                    fill=False,
                    edgecolor="navy",
                    linewidth=1.5
                )
            )

        ax.text(
            x,
            y,
            state,
            ha="center",
            va="center",
            fontsize=8
        )

    # Starting arrow
    sx, sy = positions[dfa.start_state]

    ax.annotate(
        "",
        xy=(sx - 0.25, sy),
        xytext=(sx - 0.9, sy),
        arrowprops=dict(arrowstyle="->", linewidth=1.5)
    )

    # Trap state label
    tx, ty = trap_position

    ax.text(
        tx,
        ty - 0.4,
        "TRAP STATE",
        ha="center",
        color="red",
        fontsize=9
    )

    ax.set_title(title, fontsize=14, fontweight="bold")

    ax.set_xlim(-1.2, max(1, (count - 1) * 1.5 + 0.8))
    ax.set_ylim(-1.5, 2)

    ax.axis("off")

    plt.show()


# =========================
# 4. TEST CASE GENERATOR
# =========================

def test_cases(dfa, accepted_inputs, rejected_inputs):

    results = []

    for value in accepted_inputs:

        accepted, trace = dfa.run(value)

        results.append({
            "Input": value,
            "Expected": "ACCEPT",
            "Actual": "ACCEPT" if accepted else "REJECT",
            "Result": "PASS" if accepted else "FAIL",
            "State Trace": " -> ".join(trace)
        })

    for value in rejected_inputs:

        accepted, trace = dfa.run(value)

        results.append({
            "Input": value,
            "Expected": "REJECT",
            "Actual": "ACCEPT" if accepted else "REJECT",
            "Result": "PASS" if not accepted else "FAIL",
            "State Trace": " -> ".join(trace)
        })

    return pd.DataFrame(results)


# ============================================================
# APPLICATION 1: INDIAN MOBILE NUMBER VALIDATOR
# ============================================================

print("\n\nAPPLICATION 1: MOBILE NUMBER VALIDATOR")
print("-" * 55)

# Alphabet:
# F = First digit between 6 and 9
# D = Any decimal digit
# X = Other character
#
# Language:
# L1 = Exactly 10 digits, first digit between 6 and 9

def mobile_classifier(char):

    if char in "6789":
        return "F"

    elif char in "0123456789":
        return "D"

    return "X"


mobile_states = [
    f"q{i}" for i in range(11)
] + ["qT"]

mobile_alphabet = ["F", "D", "X"]

mobile_transitions = {

    ("q0", "F"): "q1"
}

# Count remaining nine digits
for i in range(1, 10):

    mobile_transitions[(f"q{i}", "D")] = f"q{i+1}"
    mobile_transitions[(f"q{i}", "F")] = f"q{i+1}"


mobile_dfa = create_dfa(
    name="Mobile Number DFA",
    states=mobile_states,
    alphabet=mobile_alphabet,
    start="q0",
    finals={"q10"},
    trap="qT",
    transitions=mobile_transitions,
    classifier=mobile_classifier
)

print("\nMobile Number DFA Transition Table")

display(mobile_dfa.display_transition_table())

draw_dfa(
    mobile_dfa,
    "Indian Mobile Number DFA"
)

mobile_valid = [
    "9876543210",
    "9123456789",
    "8765432109",
    "7000000000",
    "6123456780"
]

mobile_invalid = [
    "5123456789",
    "987654321",
    "98765432109",
    "98A6543210",
    "+919876543210"
]

print("\nMobile Number Test Cases")

mobile_results = test_cases(
    mobile_dfa,
    mobile_valid,
    mobile_invalid
)

display(mobile_results)


# ============================================================
# APPLICATION 2: UNIVERSITY COURSE CODE VALIDATOR
# ============================================================

print("\n\nAPPLICATION 2: COURSE CODE VALIDATOR")
print("-" * 55)

# Format:
# Two uppercase letters
# Three digits
# One uppercase letter
#
# Example: CS101T
#
# Regular Expression:
# [A-Z]{2}[0-9]{3}[A-Z]

def course_classifier(char):

    if "A" <= char <= "Z":
        return "U"

    elif "0" <= char <= "9":
        return "D"

    return "X"


course_states = [
    f"r{i}" for i in range(7)
] + ["rT"]

course_alphabet = ["U", "D", "X"]

course_transitions = {

    ("r0", "U"): "r1",
    ("r1", "U"): "r2",

    ("r2", "D"): "r3",
    ("r3", "D"): "r4",
    ("r4", "D"): "r5",

    ("r5", "U"): "r6"
}


course_dfa = create_dfa(
    name="Course Code DFA",
    states=course_states,
    alphabet=course_alphabet,
    start="r0",
    finals={"r6"},
    trap="rT",
    transitions=course_transitions,
    classifier=course_classifier
)

print("\nCourse Code DFA Transition Table")

display(course_dfa.display_transition_table())

draw_dfa(
    course_dfa,
    "University Course Code DFA"
)

course_valid = [
    "CS101T",
    "AI205L",
    "MA110T",
    "EC220P",
    "IT305A"
]

course_invalid = [
    "cs101T",
    "C101T",
    "CSC01T",
    "CS10T",
    "CS1017"
]

print("\nCourse Code Test Cases")

course_results = test_cases(
    course_dfa,
    course_valid,
    course_invalid
)

display(course_results)


# ============================================================
# APPLICATION 3: HEXADECIMAL COLOUR CODE VALIDATOR
# ============================================================

print("\n\nAPPLICATION 3: HEX COLOUR CODE VALIDATOR")
print("-" * 55)

# Format:
# # followed by exactly six hexadecimal characters
#
# Example: #1A2B3C
#
# Regular Expression:
# #[0-9A-Fa-f]{6}

def hex_classifier(char):

    if char == "#":
        return "P"

    elif char in "0123456789ABCDEFabcdef":
        return "H"

    return "X"


hex_states = [
    f"c{i}" for i in range(8)
] + ["cT"]

hex_alphabet = ["P", "H", "X"]

hex_transitions = {

    ("c0", "P"): "c1"
}

# Six hexadecimal characters
for i in range(1, 7):

    hex_transitions[(f"c{i}", "H")] = f"c{i+1}"


hex_dfa = create_dfa(
    name="Hexadecimal Colour DFA",
    states=hex_states,
    alphabet=hex_alphabet,
    start="c0",
    finals={"c7"},
    trap="cT",
    transitions=hex_transitions,
    classifier=hex_classifier
)

print("\nHexadecimal Colour DFA Transition Table")

display(hex_dfa.display_transition_table())

draw_dfa(
    hex_dfa,
    "Hexadecimal Colour Code DFA"
)

hex_valid = [
    "#1A2B3C",
    "#ffffff",
    "#00FF7A",
    "#ABC123",
    "#9f8e7d"
]

hex_invalid = [
    "1A2B3C",
    "#12345",
    "#1234567",
    "#12G45Z",
    "##1234"
]

print("\nHexadecimal Colour Test Cases")

hex_results = test_cases(
    hex_dfa,
    hex_valid,
    hex_invalid
)

display(hex_results)


# ============================================================
# 5. INTERACTIVE INPUT VALIDATION
# ============================================================

def interactive_validator():

    print("\nSELECT VALIDATION APPLICATION")

    print("1. Mobile Number")
    print("2. University Course Code")
    print("3. Hexadecimal Colour Code")

    choice = input("\nEnter your choice (1/2/3): ").strip()

    validators = {
        "1": mobile_dfa,
        "2": course_dfa,
        "3": hex_dfa
    }

    if choice not in validators:

        print("Invalid selection!")
        return

    dfa = validators[choice]

    user_input = input("Enter input string: ")

    trace_df, accepted, final_state = dfa.trace_input(
        user_input
    )

    print("\nVALIDATION RESULT")

    if accepted:

        print("INPUT ACCEPTED")

    else:

        print("INPUT REJECTED")

    print("\nFinal State:", final_state)

    print("\nSTATE TRANSITION TRACE")

    if len(trace_df) > 0:

        display(trace_df)

    else:

        print("Empty input: no transitions performed.")

    result, states = dfa.run(user_input)

    print("\nComplete State Sequence:")
    print(" -> ".join(states))


# Uncomment this line to use interactive validation:
# interactive_validator()


# ============================================================
# 6. SYNTAX VS SEMANTIC VALIDATION
# ============================================================

print("\n\nSYNTAX VS SEMANTIC VALIDATION")
print("-" * 55)

comparison = pd.DataFrame({

    "Application": [
        "Mobile Number",
        "Course Code",
        "Hexadecimal Colour"
    ],

    "Syntax Validation": [
        "Exactly 10 digits, first digit 6-9",
        "Two uppercase letters, three digits, one uppercase letter",
        "# followed by six hexadecimal characters"
    ],

    "Semantic Validation": [
        "Whether the number is active or allocated",
        "Whether the course exists in the university catalogue",
        "Whether the colour is approved by the application"
    ]
})

display(comparison)


# ============================================================
# 7. FINAL TEST SUMMARY
# ============================================================

all_results = pd.concat(
    [
        mobile_results.assign(Application="Mobile Number"),
        course_results.assign(Application="Course Code"),
        hex_results.assign(Application="Hex Colour")
    ],
    ignore_index=True
)

print("\n\nFINAL AUTOMATA TEST REPORT")
print("=" * 65)

display(all_results)

total_tests = len(all_results)

passed_tests = (
    all_results["Result"] == "PASS"
).sum()

failed_tests = (
    all_results["Result"] == "FAIL"
).sum()

print("\nTOTAL TESTS:", total_tests)
print("PASSED TESTS:", passed_tests)
print("FAILED TESTS:", failed_tests)

if failed_tests == 0:

    print("\nALL TEST CASES PASSED!")

else:

    print("\nSOME TEST CASES FAILED. CHECK THE DFA TRANSITIONS.")


# ============================================================
# 8. CONCLUSION
# ============================================================

print("\n\nASSIGNMENT CONCLUSION")
print("=" * 65)

print("""
Three Deterministic Finite Automata were developed:

1. Indian Mobile Number Validator
2. University Course Code Validator
3. Hexadecimal Colour Code Validator

Each DFA contains:
- A defined alphabet
- A set of states
- An initial state
- Transition functions
- Accepting states
- A trap state

The automata validate fixed input formats by processing
characters from left to right.

Finite automata are suitable for syntax validation when
the required memory is finite.

Semantic validation, such as checking database records,
requires additional application logic.

Therefore, DFA provides a simple and efficient method
for practical input-format validation.
""")

print("END OF ASSIGNMENT 2")
print("=" * 65)
```
